```python
import json

# Function to recursively count all values in the JSON data
def count_values(data):
    if isinstance(data, dict):
        return sum(count_values(v) for v in data.values())
    elif isinstance(data, list):
        return sum(count_values(item) for item in data)
    else:
        # If it's a basic data type (e.g., int, string), count it as 1
        return 1

# Load JSON data from a file and count the values
def count_json_values(file_path):
    with open(file_path, 'r') as file:
        data = json.load(file)
        return count_values(data)

# Example usage
file_path = '../data/cleaned_tins.json'  # Replace this with the path to your file
total_values = count_json_values(file_path)
print(f"Total values in JSON: {total_values}")
```
